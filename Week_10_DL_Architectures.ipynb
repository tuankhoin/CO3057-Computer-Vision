## Stacked Convolutional Layers
![](https://storage.ghost.io/c/38/9a/389ab7af-1195-4ce6-9b1f-d9fb39ea9790/content/images/2018/04/Screen-Shot-2018-04-17-at-5.32.45-PM.png)

	- Two 3x3 conv. layers = effective receptive field of 5x5
	- Three 3x3 conv. layers = effective receptive field of 7x7


## Depthwise Separable Convolutions
![](https://editor.analyticsvidhya.com/uploads/45412spatial_seperable_convolution.png)

Given $A^{m \times n} = B^{m \times 1} \times C^{1 \times n}$. 2D convolution on A is equivalent to consecutively convolving on 2 1D filters $B,C$.

# 2. Common Architectures
---

## 2.1 AlexNet

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/alexnet.png?raw=true)

- ReLU (Rectified Linear Unit) activation function - faster training
- Training on GPU - parallelisation allows faster training (actually required 2 GPUs at the time!)
- Overlapping max pooling regions, response normalisation after ReLU - small accuracy increase
- Data augmentation - reduces overfitting
- Dropout - reduces overfitting

## 2.2 VGG

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/vgg.png?raw=true)

There are other VGG variations with more/fewer layers (e.g., VGG-19)

- Stacked 3x3 convolutional layers
	- Learn more complex features thanks to additional non-linearities
	- Fewer parameters than 1 layer with the equivalent receptive field
- Doesn't use AlexNet's response normalisation - allows faster training with only very small accuracy to drop

## 2.3 GoogLeNet (Inception)

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/googlenet.png?raw=true)

- Inception module
	- Learns features at a variety of kernel sizes/scales
- Auxillary classifiers
	- Used during training only - classify images based on early layer representations and update parameters
	- Helps with vanishing gradient problem

## Inception Module
- Choosing the right kernel size in CNNs is difficult because objects/features can appear at any scale
- Solution: use multiple kernel sizes and concatenate

- 1x1 convolutional layers reduce the number of channels (dimensionality reduction)

Deeper neural networks don't always give better performance
- Performance saturates and then decreases
- Not due to overfitting (performance can be worse on the training set)
- It should be possible to learn parameters in the deep network that would allow it to act like the small network
	- For example, some conv. layers learn identity kernels, while others learn the shallow network's kernels
- However, deep CNNs cannot learn this solution (at least, not within a reasonable training time)

⟹ Solution: Add "shortcut connections" that skip some layers

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/resnet2.png?raw=true)

## Residual Block
![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/residual_learning.png?raw=true)

- Reformulate the learning problem
	- Traditional network: input $x$, learn $H(x)$, which is the feature representation of $x$
	- Residual network: input $x$, learn $H(x) - x$, which is then added to $x$ to get $H(x)$
- Makes it easier to learn identity mapping (not losing important information)

## 2.5 MobileNets
- Lightweight architectures for mobile apps
- MobileNets uses depthwise-separable filters - 2D filters in $x$, $y$ and 1D filters over channels

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/mobilenet_architecture.png?raw=true)

- Depthwise separable convolution
	- Fewer parameters and less computation
	- Limits what kernels the model can learn - not all kernels are separable
- Smaller and faster than other architectures
	- Lower accuracy than VGG, ResNet, etc
	- But better suited for real-time applications, phones



## 2.6 U-Net
- Originally proposed for medical image segmentation
- Encoder-decoder structure with some additional features
- In decoder, the upsampled feature map is concatenated with the original features from the corresponding encoder layer

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/u-net.png?raw=true)

![](https://github.com/travisddavies/computer_vision_notes/blob/main/Images/u-net-architecture.png?raw=true)

1. We turn our image into a 1D vector
2. We have some conv blocks to process our image
3. We then do some strided conv operations to downsample
4. The bottom part is 1x1 conv operations
5. We then do a transposed conv on the beginning of the right-hand side operations to upsample the outputs, but we **also concatenate on the outputs** of the equivalent left-hand side operations as **input for the operation**
6. We repeat this process all the way back up to the top of the right-hand side output (Creating a U-Shape)
7. Finally we get a logits output as our result

- This process allows our model get a lot of context about previous information, so this information isn't lost during downsampling
- The architecture has addition information from the input itself along with the upsampled data

# 3. Modern Architectures

---

## From CNNs to Modern Backbones

* Early progress in vision focused on better **convolutions**
* Recent progress adds new ways to model:

  * **global context**
  * **long-range dependencies**
  * **scalability**
  * **efficiency**
* Today, common backbone families include:

  * CNNs
  * Transformers
  * State Space Models
  * Hybrid architectures

## 3.1 Vision Transformers (ViT)

![](https://substackcdn.com/image/fetch/$s_!RZO6!,f_auto,q_auto:good,fl_progressive:steep/https%3A%2F%2Fbucketeer-e05bbc84-baa3-437e-9518-adb32be77984.s3.amazonaws.com%2Fpublic%2Fimages%2F9c096bed-e908-49da-a716-f8a78de952a3_1358x892.png)

![](https://framerusercontent.com/images/LCmshfdtiJB1PQZWJbTkHhcBthw.webp?width=1300&height=760)

Key idea: Replace convolutional inductive bias with **attention over image patches**
* Split image into fixed-size patches
* Flatten patches into tokens
* Add positional embeddings
* Process tokens with self-attention blocks
* Final representation used for classification

Self-attention allows each token to interact with all other tokens. Good for modeling:
  * long-range relationships
  * global structure
  * object parts far apart in the image
* Scales well with large datasets and pretraining

## 3.2 Core Mechanism: Multi-Head Self-Attention

* **Self-attention**: evaluate tokens, decides which ones matter most on 1 criteria.
  - Problem: Single-minded attention may get you biased consideration

<img src="https://img.ifunny.co/images/143ec7320778e20dd064d3a8e8c1976b0850d3e96e4207c1809e1837b08f374e_1.jpg" height=300/>

~Solution: give head~
* Instead of doing this just once, **multi-head** attention does it several times in parallel ⟹ Multiple criterion
* Each **head** can learn a different kind of relationship, for example:
  * One head might relate distant patches of the same object
  * Another might focus on local texture
  * Another might capture global layout

## 3.3 Swin Transformer

![](https://substackcdn.com/image/fetch/$s_!cl_l!,f_auto,q_auto:good,fl_progressive:steep/https%3A%2F%2Fsubstack-post-media.s3.amazonaws.com%2Fpublic%2Fimages%2Ff9719e77-242c-44ac-8b3f-fef199a42d91_1521x870.png)

* A hierarchical vision Transformer
* Uses **local window attention**
* Uses **shifted windows** so information can move across windows
* Builds multi-scale features like CNN backbones

## 3.4 CNNs Advancement: ConvNeXt

Kinda Resnet, but replace the Residual block

![](https://cdn.prod.website-files.com/626ff088dc91e8769e2f3274/6418a8208ba1be75ce26640b_S1q-yFtdqw-sTA-4pq-u4s9CmR1vpuWL5rONPgldDSzwMYT4PcafxQcMdGXUKjWX0NmMZjJfdOcmtONVl5lDd2GaBKjojfhC7NWtqdSDU6IuQn0VRJE91Z_YnxKY2T0W4PsDHmgYZo5cvvEZGVuUnsc.webp)

* ConvNeXt revisited pure CNN design using lessons learned from Transformers
* Modernized ConvNets using:
  * larger kernels
  * improved normalization and training choices
  * stage design closer to Transformer backbones

* Strong performance does not require attention everywhere
* Carefully designed CNNs can still compete with Transformer models




## 3.5 State Space Models and Mamba

![](https://cdn.analyticsvidhya.com/wp-content/uploads/2026/04/image2-2.webp)

Use State Space Models (SSMs) to make attention more selective
* Transformers are powerful but expensive on long sequences
* SSMs provide another sequence modeling approach
* Mamba uses **selective state spaces**
* Can model long-range dependencies with **linear-time scaling**

Vision models based on Mamba-like ideas aim to:
  * keep long-context modeling
  * reduce memory/computation
  * compete with attention-based backbones
* A common challenge:
  * images are 2D, while Mamba is naturally sequential
* Solutions often reshape image features into scan orders or multi-dimensional state-space designs

Examples such as Mamba-ND extend Mamba-style models to multi-dimensional data including images and video. ([arXiv][5])

# 4. General Summary
---

## 4.1 Common Modern Design Ideas

* **Hierarchy**
  Multi-stage features at different resolutions
* **Multi-scale processing**
  Important for both classification and dense prediction
* **Residual connections**
  Still essential for optimization
* **Normalization layers**
  BatchNorm, LayerNorm, RMSNorm depending on architecture
* **Patch or token representations**
  Common in Transformers and some hybrid models
* **Pretraining + finetuning**
  Often more important than small architecture tweaks

## 4.2 Comparing Architecture Families

| Family      | Main strength                                 | Main weakness                                 | Typical use                                                 |
| ----------- | --------------------------------------------- | --------------------------------------------- | ----------------------------------------------------------- |
| CNN         | Efficient, strong locality bias               | Harder to model very long-range interactions  | Mobile vision, standard classification, detection backbones |
| Transformer | Strong global modeling, scales well with data | Expensive attention at high resolution        | Foundation models, large-scale vision                       |
| Mamba / SSM | Efficient long-sequence modeling              | Less mature in vision, fewer standard recipes | Long-context and emerging efficient backbones               |
| Hybrid      | Balances locality + global reasoning          | More design complexity                        | Modern practical systems                                    |

## 4.3 Architecture Trends

* Earlier models improved by designing better convolutional blocks
* Modern models improve by balancing:
  * **local feature extraction**
  * **global reasoning**
  * **training stability**
  * **compute efficiency**
* The “best” architecture depends on:
  * dataset size
  * compute budget
  * task type
  * latency constraints

# 5. How to Design a Deep Learning Model
---

## 5.1 Start With the Task

Before choosing an architecture, define:

* What is the output?

  * class label
  * bounding boxes
  * segmentation mask
  * keypoints
  * text
* What matters most?

  * accuracy
  * speed
  * memory
  * robustness
  * interpretability

Rule of thumb: Start from the task and constraints

## 5.2 Choose the Right Backbone

A backbone should match the problem. Examples:

* Small dataset, limited compute: pretrained CNN or lightweight Transformer
* Large-scale classification: ResNet / ConvNeXt / ViT / Swin
* Dense prediction: hierarchical backbones often work better
* Long-context sequential or multimodal data: Transformer or Mamba-style model may be more suitable

## 5.3 Design Questions

When designing a model, ask:

* Do I need **local detail** or **global context**?
* Do I need **multi-scale features**?
* Is my task sensitive to **resolution**?
* Is inference cost important?
* Do I have enough data to train a large model?
* Should I use pretrained weights?

## 5.4 Practical Design Strategy

A good practical recipe is:
1. Start with a strong baseline
2. Use pretrained weights if possible
3. Match input resolution to the task
4. Choose loss functions that align with evaluation
5. Add regularization only as needed
6. Tune data augmentation carefully
7. Profile both accuracy and speed

## 5.5 Typical Failure Modes in Model Design

* Model too large for dataset → overfitting
* Model too small → underfitting
* Wrong inductive bias for the task
* Evaluation metric does not match training loss
* Input preprocessing mismatch
* Poor augmentation policy
* Ignoring compute and memory limits

## 5.6 A Design Workflow

* Establish a baseline
* Diagnose the bottleneck:
  * optimization?
  * capacity?
  * data quality?
  * architecture?
* Change one major factor at a time
* Compare using a consistent validation protocol
* Prefer simpler models unless a more complex model clearly helps

## Pixels as Attributes
- Pixels are a poor space for classification
	- High-dimensional space: 256x256x3 image = 196,608 attributes
	- Irrelevant transformations (translation, lighting change, scale change, rotation, etc.) cause large changes in pixel values
- A good feature space for image recognition:
	- Is lower-dimensional - e.g., 1000s of values per image
	- Projects from the same class into a similar part of the space (images with the same class label have similar features)

## Using Pretrained Networks
- Deep learning models convert images from pixels to high-level features that are good for classification (feature embedding)
- These high-level features give good performance on a range of computer vision tasks
- Transfer learning - use features from a model trained on a large-scale task

## Transfer Learning
- **Embedding** of an input = the network's response to the input at some layer
- Extract the representation from a late layer of a model, example: CNN trained on ImageNet
	- E.g., for each image take the activations from the 4096 neurons that feed into the 1000-way ImageNet classification
- Use the neurons' activations as the attributes for a classifier of your choice (e.g., SVM, K-NN, etc.)

## Retraining Layers
- **Finetuning** = retraining layers of a pretrained CNN
- How many layers to fine tune depends on dataset size and how similar it is to ImageNet
	- More dissimilar datasets may need more retraining of lower layers
	- If dataset size is limited, training lower layers may just lead to overfitting
